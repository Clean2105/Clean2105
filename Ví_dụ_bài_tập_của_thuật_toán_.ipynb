```python
# VD 1: Tìm kiếm tuyến tính
num = list(map(int, input().split()))
search = int(input())

def linear_search(arr, target):
  Len = len(arr)
  for i in range(Len):
    if arr[i] == target:
      return True
  return False

if __name__ == '__main__':
  if linear_search(num, search):
    print(f"{search} đã tìm thấy số trong danh sách")
  else:
    print(f"{search} không tìm thấy số trong danh sách")
```

```python
# a) Hằng số.
small = list(map(int, input().split()))
large = list(map(int, input().split()))

def first_element(arr):
  return arr[0]

def get(dictionary, key):
  return dictionary[key]

print(f"Small: {first_element(small)}")
print(f"Large: {first_element(large)}")
```

```python
# b) Tuyến tính
num= list(map(int, input().split()))

def find_max(arr):
  maximum = arr[0]
  for n in arr[1:]:
    if n > maximum:
      maximum = n
  return maximum

def print_all(arr):
  for n in arr:
    print(n, end="")
  print()

print(print_all(num))
print(find_max(num))
```

```python
# VD3: Đếm các số chẵn của dãy số.
n = list(map(int, input().split()))

def count_even(arr):
  count = 0
  for i in range(len(arr)):
    if arr[i] % 2 == 0:
      count += 1
  return count

print(f"Dãy số chẵn: {count_even(n)}")
```
