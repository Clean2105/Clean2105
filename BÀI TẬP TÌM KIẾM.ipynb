```python
# Câu 1: Viết lại tìm kiếm tuần tự để có thể số lượng phép so sánh cần thiết trong quá trình tìm kiếm.
def linearsearch(A, K):
    for i in range(len(A)):
        if A[i] == K:
            return i
    return None

A = list(map(int, input().split()))
K = int(input())
res = linearsearch(A, K)
print(f"Đã tìm thấy {A[res]}") if res is not None else print("{A[res]} Không tìm thấy trong danh sách") 
```

```python
# Câu 2: Cải tiến hàm tìm kiếm nhị phân để so sánh khi tìm kiếm một khóa.
def binarysearch(A, K):
    left = 0
    right = len(A) - 1
    while left <= right:
        mid = (left + right) // 2
        if A[mid] == K:
            return mid
        else:
            if A[mid] < K:
                left = mid + 1
            else:
                right = mid - 1
    return None

A = list(map(int, input().split()))
K = int(input())
res = binarysearch(A, K)
print(f"Đã tìm thấy {A[res]}") if res is not None else print(f"{A[res]} không tìm thấy")
```
