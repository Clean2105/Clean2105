```python
# Câu 1: Thuật toán sắp xếp theo dãy (theo tự nhiên, trâu bò).
# Cho trước mảng số A. Viết chương trình, hàm sort(A) thực hiện việc sắp xếp dãy A theo thứ tự tăng dần.Sau khi thực hiện hàm, dãy A sẽ được
# sắp xếp thỏa mãn điều kiện sau:
# A[0] ⩽ A[1] ⩽ A[2] ⩽ ... ⩽ A[n - 1] 
def sort(A):
    n = len(A)
    for i in range(n - 1):
        for j in range(i + 1, n):
            if A[i] > A[j]:
                A[i], A[j] = A[j], A[i]
    return A

A = list(map(int, input().split()))
res = sort(A)
print(f"Đã sắp xếp dãy số theo thứ tự tăng dần: {res}") 
```

```python
# Câu 2: Bài toán lũy thừa.
# Viết hàm exp(a,n) tính aⁿ với a bất kỳ, n là số nguyên dương.
def exp(a, n):
    p = 1
    for i in range(n):
        p = p * a
    return p

a = int(input("Nhập số nguyên dương: "))
n = int(input("Nhập số mũ: ")) 
res = exp(a, n)
print(f"Lũy thừa: {res}") 
```
