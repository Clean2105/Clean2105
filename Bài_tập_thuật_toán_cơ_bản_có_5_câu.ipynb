```python
# 1/ Tính (n) của thuật toán sau:
# 1   C = 0
# 2   for i = 1 to n - 1
# 3       for j = i + 1 to n
# 4           for k = 1 to j
# 5               C = C + 1
# 6   return C

n = int(input())
def calculator(n):
    C = 0
    for i in range(1, n):
        for j in range(i + 1, n + 1):
            for k in range(1, j + 1):
                C += 1
    return C

res = calculator(n)
print(res)
```

```python
# 2/ Tính (n) của thuật toán sau:
# 1   C = 0
# 2   for i = 1 to n
# 3       for j = 1 to i
# 4           for k = j to i + j
# 5               C = C + 1
# 6   return C
n = int(input())
def calculator(n):
    C = 0
    for i in range(1, n + 1):
        for j in range(1, i + 1):
            for k in range(j, i + j + 1):
                C += 1
    return C

res = calculator(n)
print(res)
```
