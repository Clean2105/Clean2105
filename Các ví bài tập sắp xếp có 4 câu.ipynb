```python
# Câu 1: Viết hàm sắp xếp dãy số A theo thuật toán sắp xếp, cách thứ nhất.Chứng minh tính đúng của chương trình và thuật toán.
def insertsort(A):
    for i in range(1, len(A)):
        j = i
        while j > 0 and A[j] < A[j - 1]:
            A[j], A[j - 1] = A[j - 1], A[j]  
            j = j - 1
            
A = list(map(int, input().split()))
insertsort(A)
print("Dãy số sau khi sắp xếp:", A)
```

```python
# Câu 2: Viết hàm sắp xếp dãy số A theo thuật toán sắp xếp chọn.
def selectionsort(A):
    n = len(A)
    for i in range(n - 1):
        imin = i
        for j in range(i+1, n):
            if A[j] < A[imin]:
                imin = j
        A[i], A[imin] = A[imin], A[i]

A = list(map(int, input().split()))
selectionsort(A)
print("Dãy số sau khi sắp xếp: ", A) 
```

```python
# Câu 3: Viết chương trình mô tả thuật toán sắp xếp chọn theo cách khác, trong đó,, vòng lặp tại dòng lệnh 3 sẽ duyệt dãy từ phải 
# sáng trái (Từ cuối về đầu dãy).
def selectionsort(A):
    n = len(A)
    for i in range(n - 1, 0, -1):
        imax = i  
        for j in range(i - 1, -1, -1):
            if A[j] > A[imax]:
                imax = j  
        A[i], A[imax] = A[imax], A[i]

A = list(map(int, input().split()))
selectionsort(A)
print(f"Dãy số sau khi sắp xếp: {A}")
```

```python
# Câu 4: Viết hàm sắp xếp dãy số A theo thuật toán sắp xếp nổi bọt.
def bubblesort(A):
    n = len(A)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if A[j] > A[j + 1]:
                A[j], A[j + 1] = A[j + 1], A[j] 

A = list(map(int, input().split()))
bubblesort(A)
print(f"Dãy số đầu tiên sau khi sắp xếp: {A}") 
```
