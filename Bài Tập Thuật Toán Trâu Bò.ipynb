```python
# Đề: Viết chương trình thuật toán đầu ra là bộ giá trị: left, right, Maxlen và in ra màn hình: (left, right, Maxlen).
def Maxlen_Array(A):
    n = len(A)
    imax = 0
    jmax = 0
    lenmax = 1
    for i in range(n):
        j = i
        length = 1
        while j < n - 1 and A[j] <= A[j + 1]:
            j += 1
            length += 1
        if lenmax < length:
            lenmax = length
            imax = i
            jmax = j
    return imax, jmax, lenmax

A = list(map(int, input().split()))
res = Maxlen_Array(A)
print(res)
```
