```python
# Câu 1: Viết chương trình mô phỏng trò chơi tháp Hà Nội.
C = ["A", "B", "C"]
def Hanoi(n, i, j, k):
    if n == 0:
        return
    Hanoi(n - 1, i, k, j)
    
    print(f"Chuyển đĩa {n} từ cọc {C[i]} sang cọc {C[j]}")   
    
    Hanoi(n - 1, k, j, i)
    
    print(f"Chuyển đĩa {n} từ cọc {C[i]} sang cọc {C[j]}")

n, i, k, j = map(int, input().split())
res = Hanoi(n, i, j, k) 
print(res) 
```

```python
# Câu 2: Giả sử a, b là 2 số nguyên không âm và không đồng thời bằng 0. Chứng minh thuật toán sau tính đúng UCLN gcd(a, b).
# 1| Input: a, b
# 2| k = max(a, b)
# 3| while a mob k ≠ 0 or b mob n ≠ 0
# 4|     k = k - 1
# 5| return k 
def UCLN_gcd(a, b):
    k = max(a, b)
    while (a % k != 0) or (b % k != 0):
        k -= 1
    return k

a, b = map(int, input().split())
res = UCLN_gcd(a, b)
print(res) 
```
