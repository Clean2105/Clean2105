```python
# Câu 1: Mô hình ngăn xếp (stack) chuẩn sử dụng cấu trúc tự định nghĩa.
class Stack:
    def __init__(self):
        self.items = []

def isEmpty(S):
    return len(S.items) == 0

def push(S, x):
    S.items.append(x)

def pop(S):
    if isEmpty(S):
        raise RuntimeError("Rỗng!")
    topIdx = len(S.items) - 1
    x = S.items[topIdx]
    del S.items[topIdx]
    return x
    
S = Stack()

print("1. Kiểm tra ngăn xếp mới tạo:")
print("Ngăn xếp có rỗng không?", isEmpty(S))

print("\n2. Thực hiện push các số có thể vào ngăn xếp...")
push(S, 10)
push(S, 20)
push(S, 30)
push(S, 40)
push(S, 50)

print("Ngăn xếp hiện tại có rỗng không?", isEmpty(S))

print("\n3. Thực hiện pop các phần tử ra khỏi ngăn xếp:")
print("Phần tử lấy ra lần 1:", pop(S))
print("Phần tử lấy ra lần 2:", pop(S))
print("Phần tử lấy ra lần 3:", pop(S))

print("\n4. Thử pop tiếp khi ngăn xếp đã rỗng:")
try:
    print(pop(S))
except RuntimeError as e:
    print(f"Lỗi bắt được: {e} (Không thể pop vì ngăn xếp không còn phần tử nào)")
```

```python
# Câu 2: Cấu trúc câu nhị phân(Binary tree).
class Node:
    def __init__(self, key, p = None, left = None, right = None):
        self.key = key
        self.p = p
        self.left = left
        self.right = right
class Tree:
    def __init__(self, key = None):
        if key == None:
            self.root = None
        else:
            self.root = Node(key)

def preOrder(node):
    if node is not None:
        print(node.key, end=" ")
        preOrder(node.left)
        preOrder(node.right)

def inOrder(node):
    if node is not None:
        inOrder(node.left)
        print(node.key, end=" ")
        inOrder(node.right)

def postOrder(node):
    if node is not None:
        postOrder(node.left)
        postOrder(node.right)
        print(node.key, end=" ")

T = Tree(10)

node_left = Node(20, p=T.root)
node_right = Node(30, p=T.root)

T.root.left = node_left
T.root.right = node_right

print("Nút gốc (Root):", T.root.key)
print("Con bên trái của Root:", T.root.left.key)
print("Con bên phải của Root:", T.root.right.key)
print("Cha của nút con bên trái:", node_left.p.key)
```
