```python
# Câu 1: Thuật toán phân đoạn gốc của Hoare, sử dụng pivot:
# a/ Phải.
# b/ Trái.

# a/ Phải.
def partition(A, left, right):
    pivot = A[right]
    i = left
    j = right - 1
    while i <= j:
        while i <= right and A[i] <= pivot:
            i += 1
        while j >= left and A[j] > pivot:
            j -= 1
        if i < j:
            A[i], A[j] = A[j], A[i]  
            i += 1
            j -= 1
    A[i], A[right] = A[right], A[i]
    return i

A = list(map(int, input().split()))
left = 0
right = len(A) - 1
print(f"Mảng sau khi partition: {A}")
print(f"Bên phải: {partition(A, left, right)}") 
```

```python
# b/ Trái
def partition(A, left, right):
    pivot = A[right]
    i = left + 1
    j = right
    while i <= j:
        while i <= right and A[i] <= pivot:
            i += 1
        while j > left and A[j] > pivot:
            j -= 1
        if i < j:
            A[i], A[j] = A[j], A[i]  
            i += 1
            j -= 1
    A[left], A[j] = A[j], A[left]
    return j

A = list(map(int, input().split()))
left = 0
right = len(A) - 1
print(f"Mảng sau khi partition: {A}")
print(f"Bên trái: {partition(A, left, right)}") 
```

```python
# Câu 2: Thiết lập thuật toán sắp xếp sau, gọi là ThreeSort(A, i, j).
def ThreeSort(A, i, j):
    if i >= j:
        return
    if A[i] > A[j]:
        A[i], A[j] = A[j], A[i]
    if j - i + 1 > 2:
        m = (j - i + 1) // 3
        
        ThreeSort(A, i, j - m)      
        ThreeSort(A, i + m, j)      
        ThreeSort(A, i, j - m)      
A = list(map(int, input().split()))
ThreeSort(A, 0, len(A) - 1)
print(f"Đã sắp xếp: {A}")
```
