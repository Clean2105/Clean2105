```python
# Câu 1: Sinh dãy nhị phân độ dài n, phương án không đệ quy.
def GenBinary(n):
    A = [0] * n
    
    while True:
        print(A)  
        
        i = n - 1
        while i >= 0 and A[i] == 1:
            A[i] = 0  
            i -= 1
        if i < 0:
            break
        A[i] = 1

n = int(input())
GenBinary(n)
```

```python
# Câu 2: Sinh dãy nhị phân có độ dài n, phương án đệ quy.
def Binary(A, k, n):
    for i in range(2):
        A[k] = i  
        
        if k < n - 1:
            Binary(A, k + 1, n) 
        else:
            print(A)  

n = int(input())
A = [0] * n
Binary(A, 0, n)
```
