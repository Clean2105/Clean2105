```python
# Câu 1: Viết chương trinh bài toán đổi tiền (CMP - Change Making Problem).
def ChangeMaking(C, n):
    count = 0
    k = len(C) - 1
    while n > 0:
        count = count + n//C[k] 
        n = n % C[k]
        k = k - 1
    return count 

C = list(map(int, input().split()))
n = int(input()) 
res = ChangeMaking(C, n)
print(f"Bài toán đổi tiền: {res}") if res is not None else print("Không có số tiền trong thuật toán tham lam!") 
```

```python
# Câu 2: Giải bài toán xếp ba lô phân số bằng chiến lược tham lam.
def indexSorting(A, h):
    n = len(A)
    ID = list(range(n))
    if h == 1:
        ID = sorted(ID, key=lambda x: A[x], reverse = False)
    else:
        ID = sorted(ID, key=lambda x: A[x], reverse = True) 
    return ID     

def FractionalKnapsack(W, weights, values):
    n = len(weights)
    ratio = [values[i] / weights[i] for i in range(n)]
    
    sorted_indices = indexSorting(ratio, h=0)
    
    total_value = 0.0  
    fractions = [0] * n 
    for i in sorted_indices:
        if weights[i] <= W:
            W -= weights[i]
            total_value += values[i]
            fractions[i] = 1
        else:
            fractions[i] = W / weights[i]
            total_value += values[i] * fractions[i]
            W = 0  
            break 
    return total_value, fractions

W = int(input())
weights = list(map(int, input().split()))
values = list(map(int, input().split()))
    
max_val, fractions = FractionalKnapsack(W, weights, values)
    
print(f"Giá trị lớn nhất là: {max_val}")
print("Tỷ lệ lấy của từng vật phẩm:")
for i in range(len(weights)):
    print(f"- Vật phẩm {i+1} (w={weights[i]}, v={values[i]}): Lấy {fractions[i]*100:.1f}%")
```

```python
# Câu 3: Bài toán chọn công việc (ASP - Activity Selection Problem).
def indexSorting(A, h):
    n = len(A)
    ID = list(range(n))
    if h == 1:
        ID = sorted(ID, key=lambda x: A[x], reverse = False)
    else:
        ID = sorted(ID, key=lambda x: A[x], reverse = True) 
    return ID    

def GreedyASP(s, f):
    n = len(s)
    h = indexSorting(f, 1)
    A = []
    k = 0
    A.append(h[k])
    for i in range(1, n):
        if s[h[i]] >= f[A[k]]:
            k += 1
            A.append(h[i])
    return A
s = list(map(int, input().split()))
f = list(map(int, input().split()))
res = GreedyASP(s, f)
print(f'Bài toán chọn công việc: {res}') 
```
