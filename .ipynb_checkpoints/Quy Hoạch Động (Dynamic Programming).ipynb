```python
# Câu 1: Viết chương trình tính số Catalan thứ n.
def Calatan(n):
    C = [0] * (n + 1)
    C[0] = 1
    for i in range(1,n+1):
        for j in range(i):
            C[i] = C[i] + C[j]*C[i-1-j]
    return C[n]

n = int(input())
r = Calatan(n)
print(f'Calatan = {r}') 
```

```python
# Câu 2: Hệ số hoán vị P(n, k), 0 ⩽ k ⩽ n.
def sinhMATRIX(m, n):
    A = []
    for i in range(m):
        A.append([0]*m)
    return A  
    
def PermCoeff(n):
    p = sinhMATRIX(n + 1, n + 1)
    for k in range(n + 1):
        p[k][0] = 1
        for m in range(1, n+1):
            for j in range(m+1):
                p[m][j] = p[m-1][j] + j * p[m-1][j-1]
    return p

n = int(input())
r = PermCoeff(n)
print(f" Hệ số hoán vị: {r}") 
```
