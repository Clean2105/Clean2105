```python
# Câu 1: Bài toán đơn điệu tăng cực đại(LIS - Longest Increasing Subsequence). 
def printLIS(A, L):
    n = len(A)  
    Idx = 0
    for k in range(n):
        if L[k] > L[Idx]:
            Idx = k
    k = Idx
    kq = []
    kq.append(A[Idx])
    while k >= 0: 
        j = k - 1
        while j >= 0 and (L[k] != 1 + L[j] or A[k] <= A[j]):
            j -= 1
        if j >= 0:
            kq.append(A[j])
        k = j  
    kq.reverse()
    return kq

A = list(map(int, input().split()))
L = list(map(int, input().split()))
r = printLIS(A, L)
print(f"LIS là: {r}") 
```

```python
# Câu 2: Viết hàm tính số Fibonacci thứ n và sử dụng thuật toán quy hoạch động.
def Fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        F = [0] * (n + 1)
        F[1] = 1
        for k in range(2, n+1):
            F[k] = F[k - 1] + F[k - 2]
        return F[n]
n = int(input())
r = Fib(n)
print(f"Fib = {r}") 
```
