```python
# Câu 1: Xây dựng hàm sắp xếp các xâu ký tự sao cho các xâu được tạo thành từ cùng một tập ký tự nằm liền kề nhau.
# Ví dụ: "hello" và "lohel" là hai từ cùng tạo thành từ một tập ký tự.
def StringMatching(T, P):
    n = len(T)
    m = len(P)
    if n != m:
        return -1
    list_P = list(P)
    for k in range(n):
        j = 0
        matched = False
        while j < len(list_P):
            if T[k] == list_P[j]:
                list_P.pop(j)
                matched = True
                break
            j = j + 1
        if not matched:
            return -1
    return 0

T = input("Nhập chuỗi văn bản lớn T: ")
P = input("Nhập chuỗi con cần tìm P: ")
res = StringMatching(T, P)
if res == 0:
    print(f'Chuỗi "{T}" và "{P}" đã tạo thành công')
else:
    print(f'Chuỗi "{T}" và "{P}" KHÔNG tạo thành công')
```

```python
# Câu 2: Cho một mảng các số nguyên được sắp xếp theo thứ tự tăng dần, thực hiện xoay các phần tử trong mảng với số lần bất kỳ.
# Ví dụ: Mảng ban đầu 1,2,3,4,5,6 xoay lần thứ nhất ta được 2,3,4,5,6,1 xoay lần thứ 2 ta được 3,4,5,6,1,2
# Xây dựng một thuật toán với thời gian cỡ O(logn) để tìm chỉ số phần tử thứ k trong mảng xoay.
# Ví dụ: Với mảng đã xoay 3,4,5,6,1,2 thì chỉ số phần tử thứ 2 là 5.
def Sorthard(A, B):
    S = [[A[i], B[i]] for i in range(len(A))]
    S.sort(key = lambda x: x[1])
    C = [x[0] for x in S]
    D = [x[1] for x in S]
    return C, D

A = list(map(int, input("Nhập mảng xoay A: ").split()))

B = list(range(len(A)))
C, D = Sorthard(A, B)
k = int(input("Nhập chỉ số k cần tìm: "))

if 0 <= k < len(A):
    print(f"Mảng ban đầu A: {A}")
    print(f"Phần tử ở chỉ số thứ {k} là: {A[k]}")
else:
    print(f"Chỉ số {k} không hợp lệ!") 
```

```python
# Câu 3: Cho một dãy số bất kỳ, xây dựng hàm để tìm và trả về các cặp số có sự khác biệt lớn nhất và nhỏ nhất
def bubblesort_a(A):
    n = len(A)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if A[j] > A[j + 1]:
                A[j], A[j + 1] = A[j + 1], A[j]
    return A

def bubblesort_b(A):
    n = len(A)
    for i in range(1, n):
        for j in range(n - 1, i - 1, -1):
            if A[j] > A[j - 1]:
                A[j], A[j - 1] = A[j - 1], A[j]
    return A
A = list(map(int, input("Nhập dãy số: ").split()))
res_A = bubblesort_a(A.copy()) 
res_B = bubblesort_b(A.copy())

print(f"Dãy số xuôi (a): {res_A}")
print(f"Dãy số ngược (b): {res_B}")

n = len(res_A)
if n >= 2:
    max_pair = (res_A[0], res_A[-1])
    min_diff = abs(res_A[1] - res_A[0])
    min_pair = (res_A[0], res_A[1])
    
    for i in range(1, n - 1):
        diff = abs(res_A[i + 1] - res_A[i])
        if diff < min_diff:
            min_diff = diff
            min_pair = (res_A[i], res_A[i + 1])
            
    print(f"LỚN NHẤT: {max_pair}")
    print(f"NHỎ NHẤT: {min_pair}")
```

```python
# Câu 4: Xây dựng hàm tìm và trả về phần tử được lặp lại nhiều nhất trong dãy số ban đầu.
def findsort(A, B):
    n = len(A)
    A, B = A.copy(), B.copy() 
    for i in range(1, n):
        valA = A[i]
        valB = B[i]
        j = i - 1
        while j >= 0 and B[j] < valB:
            A[j + 1] = A[j]
            B[j + 1] = B[j]
            j = j - 1
        A[j + 1] = valA
        B[j + 1] = valB
    return A, B

str_B = input("Nhập dãy số ban đầu: ")
A_origin = list(map(int, str_B.split()))

if not A_origin:
    print("Dãy số rỗng!")
else:
    value_A = []
    for x in A_origin:
        if x not in value_A:
            value_A.append(x)
            
    fre_B = [A_origin.count(x) for x in value_A]
    sorted_value_A, sorted_fre_B = findsort(value_A, fre_B)
    
    the_most = sorted_value_A[0]
    find_count_max = sorted_fre_B[0]
    
    print(f"Dãy số ban đầu: {A_origin}")
    print(f"Phần tử xuất hiện nhiều nhất là {the_most} (lặp lại {find_count_max} lần).")
```
