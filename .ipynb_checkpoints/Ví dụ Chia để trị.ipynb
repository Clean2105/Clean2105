```python
# Câu 1: Thiết lập thuật toán tìm kiếm nhị phân bằng chia để trị, phân tích độ phức tạp thời gian của thuật toán này.
def binSearch(A, left, right, C):
    if left <= right:
        mid = (left + right) // 2
        if A[mid] == C:
            return mid
        else:
            if A[mid] < C:
                return binSearch(A, mid+1, right, C)
            else:
                return binSearch(A, left, mid-1, C)
    else:
        return None

A = list(map(int, input().split()))
A.sort()
C = int(input())
left = 0
right = len(A) - 1
res = binSearch(A, left, right, C)
print(A[res]) if res is not None else print(f"Không tìm thấy {A[res]}") 
```

```python
# Câu 2: Viết lời giải của bài toán lũy thừa exp(a,n) = aⁿ, theo thuật toán trên nhưng không đệ quy.
def exp(a, n):
    x = 1
    while n > 0:
        if n % 2 == 0:
            x = x * a
        x = x * x
        n = n // 2
    return x
n = int(input())
a = int(input())
res = exp(a, n)
print(f"Lũy thừa = {res}") 
```

```python
# Câu 3: Thiết lập chương trình theo thuật toán nhanh Karatsuba.
def fasstnhanh(x, y, n):
    if n == 1:
        return x * y
    else:
        m = n // 2
        p = 10 ** m
        a, b = x // p, x % p
        c, d = y // p, y % p
        e = fasstnhanh(a, c, m)
        f = fasstnhanh(b, d, m)
        g = fasstnhanh(a-b, c-d, m)
        return p**2*e + p*(e+f-g) + f
x, y, n = map(int, input().split())
res = fasstnhanh(x, y, n)
print(res) 
```

```python
# Câu 4: Cho trước dãy số A. Phần tử của dãy A được gọi là yên ngựa(peak element) nếu nó không nhỏ hơn các phần tử đứng bên cạnh.
# Nếu có 1 phân tử thì vẫn coi phần tử này là yên ngựa.
def peak(A):
    left = 0
    right = len(A) - 1
    while left < right:
        mid = (left + right) // 2
        if A[mid] > A[mid+1]:
            right = mid
        else:
            left = mid + 1
    return left, A[left]
A = list(map(int, input().split()))
print(f'Yên ngựa:{peak(A)}')
```
