```python
# b) 1, 2, 6, 15.
def changeMoney_2(C1, n1):
    m = len(C1)
    X1 = [0] * m
    k = m - 1 
    while n1 > 0 and k >= 0:
        X1[k] = n1 // C1[k]  
        n1 = n1 % C1[k]      
        k -= 1
    return X1

def Shownkq_2(n1, X1, C1): 
    print(f"{n1} có thể đổi thành {sum(X1)} tờ tiền:")
    for i in range(len(C1)):
        if X1[i] > 0:
            print(f"- {X1[i]}  => tờ tiền: {C1[i]}")
C1 = list(map(int, input().split()))
n1 = int(input())
res_X1 = changeMoney_2(C1, n1)
Shownkq_2(n1, res_X1, C1)
```

```python
# Câu 2: Chứng minh rằng hệ tiền C = (1, 5, 10, 25) là chuẩn tắc.
def changeMoney(M, C):
    n = len(C)
    x = [0] * n
    k = n - 1
    while k >= 0 and M > 0:
        x[k] = M // C[k]
        M = M % C[k]
        k -= 1
    return x

C = list(map(int, input().split()))
M = int(input())
r = changeMoney(M, C)
print(f"Hệ tiền: {r}") 
```
