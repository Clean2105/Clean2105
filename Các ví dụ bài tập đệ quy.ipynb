```python
# Câu 1: Viết công thức truy hồi của các hàm sau và lập trình để thực hiện các hàm này bằng đệ quy.
# a) Hàm factorial(n) = n! (tính n giai thừa). 
# b) Hàm exp(a,n) = aⁿ (lũy thừa).
# c) Hàm S(n) = 1 + 2 + ... + n.
# d) Hàm Sqrt(n) = 1² + 2² + ... + n².
# e) Hàm Seven(n) = tổng các số tự nhiên chẵn không vượt quá n.
# f) Hàm Sodd(n) = tổng các số tự nhiên lẻ không vượt quá n.

# a) Hàm factorial(n) = n! (tính n giai thừa). 
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
n = int(input())
print(factorial(n)) 
```

```python
# b) Hàm exp(a,n) = aⁿ (lũy thừa).
def exp(a, n):
    if n == 0:
        return 1
    else:
        return a * exp(a, n - 1)

n = int(input())
a = int(input())
print(exp(a, n)) 
```

```python
# c) Hàm S(n) = 1 + 2 + ... + n.
def S(n):
    if n == 0:
        return 0
    else:
        return n + S(n - 1)

n = int(input())
print(S(n)) 
```

```python
# d) Hàm Sqrt(n) = 1² + 2² + ... + n².
def Sqrt(n):
    if n == 0:
        return 0
    else:
        return n * n + Sqrt(n - 1)

n = int(input())
print(Sqrt(n)) 
```

```python
# e) Hàm Seven(n) = tổng các số tự nhiên chẵn không vượt quá n.
def Seven(n):
    if n == 0:
        return 0
    else:
        if n % 2 == 0:
            return n + Seven(n - 1)
        else:
            return Seven(n - 1)

n = int(input())
print(Seven(n))
```

```python
# f) Hàm Sodd(n) = tổng các số tự nhiên lẻ không vượt quá n.
def Sodd(n):
    if n == 0:
        return 0
    else:
        if n % 2 == 0:
            return Sodd(n - 1)
        else:
            return n + Sodd(n - 1)

n = int(input())
print(Sodd(n))
```

```python
# Câu 2: Dãy số Fibonacci:
# a) F₀ = 0 F₁ = 1.
# b) Fₙ = Fₙ₋₁ + Fₙ₋₂ với n > 1
def Fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return Fibonacci(n - 1) + Fibonacci(n - 2)

n = int(input())
res = Fibonacci(n)
print(f"Dãy số là: {res}") 
```

```python
# Câu 3: Dãy số Pell:
# a) P₀ = 0, P₁ = 1
# b) Pₙ = 2Pₙ₋₁ + Pₙ₋₂ với n > 1
def Pell(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return 2 * Pell(n - 1) + Pell(n - 2) 

n = int(input())
res = Pell(n)
print(F"Dãy số là: {res}") 
```
