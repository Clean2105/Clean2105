```python
# VD1: Bài toán tìm kiếm tuần tự chuẩn trên mảng 2 chiều.
def MatrixSearch(A, K):
    m = len(A)
    if m == 0:
        return None
    n = len(A[0])

    for i in range(m):
        for j in range(n):
            if A[i][j] == K:
                return i, j
    return None

m, n = map(int, input().split())
print(f"Nhập ma trận {m} x {n} mỗi hàng cột:")

A = []
for _ in range(m):
    row = list(map(int, input().split()))
    A.append(row)

K = int(input("Nhập K để tìm số: "))

res = MatrixSearch(A, K)
print(res) if res is not None else print("Không tìm thấy")
```

```python
# VD2: Cho trước 2 xâu: xâu T gọi là xâu gốc, xâu P gọi là xâu mẫu.Cần tìm
# vị trí xâu mẫu P trong xâu gốc T.Nếu tìm thấy sẽ trả về vị trí đầu tiên xuất
# hiện xâu mẫu P trong T, nếu không thấy thì trả về -1.
# Ví dụ với xâu T = "ababababa", P = "aba" thì kết quả là 0 vị trí gặp đầu tiên
# P trong T.
def StringMatch(T, P):
  n = len(T)
  m = len(P)
  for i in range(n - m + 1):
    j = 0
    while j < m and T[i + j] == P[j]:
      j = j + 1
    if j == m:
      return i
  return -1

m, n = map(int, input().split())
print(f"Nhập xâu giữa {m} và {n}:")
T = input(f"Nhập xâu của {m}: ")
P = input(f"Nhập xâu của {n}: ")
res = StringMatch(T, P)
print(res)

```

```python
# VD 3: Phần tử A[i] trong dãy số A được gọi là phần tử yên ngựa 0 < i < len(A) - 1
# và A[i] thỏa mãn một trong 2 điều kiện sau:
# (1) A[i - 1] < A[i] > A[i + 1]
# (2) A[i - 1] > A[i] < A[i + 1]
def PeakCount(A):
  if len(A) < 3:
    return 0
  Count = 0
  for i in range(1, len(A) - 1):
    if A[i - 1] < A[i] > A[i + 1] or A[i - 1] > A[i] < A[i + 1]:
      Count = Count + 1
  return Count

A = list(map(int, input().split()))
res = PeakCount(A)
print(res)
```

```python
# VD 4: Cho trước số K, tìm tất cả chỉ số i sao cho A[i] = K. Hàm cần trả về 2 chỉ
# số i, j sao cho A[i] = A[i + 1] = ... = A[j] = K.Nếu Không tìm thấy trả về None.
def BinarySearch(A, K):
    left = 0
    right = len(A) - 1
    while left <= right:
        mid = (left + right) // 2
        if A[mid] == K:
            i = mid
            while i >= left and A[i] == K:
                i = i - 1
            j = mid
            while j <= right and A[j] == K:
                j = j + 1
            return i + 1, j - 1
        elif A[mid] < K:
            left = mid + 1
        else:
            right = mid - 1
    return None

if __name__ == "__main__":
    print("Nhập số lượng phần tử N:")
    N = int(input())
    print(f"Nhập {N} phần tử :")
    A = list(map(int, input().split()))

    print("Nhập số K cần tìm:")
    K = int(input())
    res = BinarySearch(A, K)
    print(f"Đã tìm thấy {res}") if res is not None else print("Không tìm thấy!")

```

```python
# VD 5: Cho trước dãy số nguyên A được sắp xếp tăng dần thực sự:
# A[0] < A[1] < ... < A[n - 1]
# Cho trước số tự nhiên K với điều kiện: A[0] < K ⩽ A[n - 1].
# Viết chương trình tìm và trả về số K sao cho: A[n - 1] < K ⩽ A[K]
def bSearch(A, K):
  left = 0
  right = len(A) - 1
  ans = None
  while left <= right:
    mid = (left + right) // 2
    if A[mid] >=  K:
      ans = mid
      right = mid - 1
    else:
      left = mid + 1
  return ans

if __name__ == "__main__":
  A = list(map(int, input().split()))
  K = int(input())
  res = bSearch(A, K)
  print(A[res]) if res is not None else print("Không tìm thấy!")
```
